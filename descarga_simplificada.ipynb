# Descarga de subtítulos VTT simplificada
Este notebook descarga únicamente subtítulos `.vtt` de YouTube y los exporta a un CSV limpio.

```python
import yt_dlp
import os

# Parámetros
FUENTE = "https://www.youtube.com/@IvanCepedaCastro/videos"
FECHA_INICIO = "20250401" 
FECHA_FIN    = "20260429" 
OUTPUT_DIR = "subtitulos"

os.makedirs(OUTPUT_DIR, exist_ok=True)

# 
def filtro_fechas(info):
    fecha = info.get('upload_date')
    
    if not fecha: return None
    
    # Videos antiguos
    if fecha < FECHA_INICIO:
        print(f"\n[Freno] Video del {fecha} alcanzado. Deteniendo descarga por completo...")
        raise KeyboardInterrupt("Cierre por fecha")
    
    # Videos nuevos
    if fecha > FECHA_FIN:
        return "Fuera de rango (muy nuevo)"
    
    return None

ydl_opts = {
    "skip_download": True,
    "writesubtitles": True,
    "writeautomaticsub": True,
    "subtitleslangs": ["es", "es-419", "es-CO", "es-MX"],
    "subtitlesformat": "vtt",
    "match_filter": filtro_fechas,
    "outtmpl": os.path.join(OUTPUT_DIR, "%(upload_date)s_%(id)s_%(title).60s.%(ext)s"),
    "ignoreerrors": True, # Salta videos privados
}

print(f"Inicio de descargas...")

try:
    with yt_dlp.YoutubeDL(ydl_opts) as ydl:
        ydl.download([FUENTE])
except KeyboardInterrupt:
    print("\n Proceso detenido exitosamente")

print("\n--- Tarea finalizada ---")
```

```python
import re
import glob
import pandas as pd
import csv

def limpieza_vtt(ruta: str) -> str:
    try:
        with open(ruta, "r", encoding="utf-8", errors="ignore") as f:
            contenido = f.read().replace('\r\n', '\n').replace('\r', '\n')
    except:
        return ""

    lineas = contenido.split('\n')
    limpias = []
    for l in lineas:
        l = l.strip()
        if not l or re.match(r"^(WEBVTT|Kind:|Language:|NOTE|\d{2}:\d{2}|-->|\d+$)", l):
            continue
        l = re.sub(r"<[^>]+>", "", l)
        l = re.sub(r"\[.*?\]", "", l)
        if l.strip():
            limpias.append(l.strip())

    texto_unido = []
    for l in limpias:
        if not texto_unido or l != texto_unido[-1]:
            texto_unido.append(l)
    
    texto_final = " ".join(texto_unido)
    texto_final = "".join(c for c in texto_final if c.isprintable())
    texto_final = re.sub(r"\s+", " ", texto_final).strip()
        
    return texto_final

corpus = []
archivos = sorted(glob.glob(os.path.join(OUTPUT_DIR, "*.vtt")))

for ruta in archivos:
    nombre = os.path.basename(ruta)
    try:
        
        fecha = nombre[:8]
        video_id = nombre[9:20]
        resto = nombre[21:]
        titulo = resto.split(".vtt")[0].replace(".es", "").replace(";", " ")
    except:
        continue

    texto = limpieza_vtt(ruta)
    if texto:
        corpus.append({
            "video_id": video_id,
            "fecha": fecha,
            "titulo": titulo,
            "fuente": f"https://youtu.be/{video_id}",
            "texto": texto,
            "palabras": len(texto.split())
        })

df = pd.DataFrame(corpus)

if not df.empty:
    # Exportar con opciones estrictas
    df.to_csv("corpus_youtube.csv", index=False, sep=";", encoding="utf-8-sig", quoting=csv.QUOTE_ALL)
    print(f"¡Completado! {len(df)} archivos exportados correctamente a 'corpus_youtube.csv'.")
else:
    print("No se encontraron subtítulos para exportar.")
```
